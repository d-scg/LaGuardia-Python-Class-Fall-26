Q. Filter your data

```python
#Filtering based on a single condition: Select all the cars with less than 25 miles per gallon [mpg] (less fuel efficient)
#Uses Boolean Indexing-->matching rows True for a condition (being less than 25)
df_analysis[
    df_analysis['mpg']<25
    ]
```

```python
#Select The Mercedes-benz cars that are less fossil efficients (mpg<25)
df_analysis[
    (df_analysis['mpg'] < 25) &
    (df_analysis['car_type'].str.contains('merc', case=False, na=False))
]
```

```python
#Filtering using multiple conditions:What are the most fuel efficient cars with smaller engines? Select cars with less than 5 cylinders and more than 30 mpg
#Uses boolean indexing--> matching rows True for more than one condition introduced by AND (&)
df_analysis[
    (df_analysis['cyl']<5) &
    (df_analysis['mpg']>30)
    ]
```
