#Data cleaning of dataset subset

```python
df_queens_clean=df_queens_hs.copy()
```

```python
# Inspecting the dataset for placeholders values
df_queens_clean['Number Scoring 65 or Above'].value_counts()
```

```python
#Checking which placeholders values are recorded
df_queens_clean['Mean Score']= df_queens_clean['Mean Score'].astype(float)
```

```python
placeholder_counts= (df_queens_clean== "s").sum()
print(placeholder_counts)
```

```python
#Calculate percentage of "s" placeholder across the dataset subset
percentage_placeholder = (placeholder_counts / len(df_queens_clean)) * 100
print(percentage_placeholder)
```

```python
print(len(df_queens_clean))
```

```python
#Transform "s" into a missing values in all rows where present
```

```python
score_columns = [
    "Mean Score",
    "Number Scoring Below 65",
    "Percent Scoring Below 65",
    "Number Scoring 65 or Above",
    "Percent Scoring 65 or Above",
    "Number Scoring 80 or Above",
    "Percent Scoring 80 or Above"
]
```

```python
df_queens_clean[score_columns] = df_queens_clean[score_columns].replace("s", pd.NA)
```

```python
df_queens_clean.info()
```

```python
df_queens_clean.isna().sum()
```

```python
df_queens_clean
```

```python
#Convert string score columns into numerical
df_queens_clean[score_columns] = df_queens_clean[score_columns].apply(pd.to_numeric)
```

###Considerations on missing values
Missing values were present in the dataset. The value "s" was identified as a missing value and was replaced with pd.NA, allowing the relevant score columns to be converted to numeric data types. Missing observations were retained rather than replaced. Regents exams aren't represented in exactly the same number of schools. The missingness is fairly similar across the three years (2015,2016,2017), but varies considerably by school and by exam.

#Conclusions to the Research question:
 The dataset contains Regents examination results for New York City public schools. Each row represents a reported set of Regents results for a school, exam, and year, rather than an individual student record.
 The analysis answer the question: How does Jamaica Gateway to the Sciences compare with other Queens high schools in Common Core Algebra performance from 2015 to 2017? Jamaica Gateway's average reported Mean Score was higher than the comparison group's in all three years (62.6 in 2015,about 1.03 points higher than the comparison group; 67.7 in 2016, 0.43 point higher than the comparison group, 69.5 in 2017, 1.30points higher than the comparison group). The data shows an increase over time for both group.The reported Mean Scores of Jamaica gateway were more spread out in 2017 than in 2015, but the comparison group has considerably more variation in reported Mean Scores than Jamaica Gateway (SD>6).
